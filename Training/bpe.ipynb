```python
from data import MINI_TEXTUAL_DATA
from rich import print

data = MINI_TEXTUAL_DATA
```

## Pretokenization

```python
import re

tokens = re.findall(r"\w+|[^\w\s]", data)
len(tokens)
```

## Creating char voc

```python
corpus = []
for token in tokens:
    chars = list(token)
    chars.append("</w>")
    corpus.append(chars)
    
print(corpus[:10])
```

## Making pair

```python
from collections import defaultdict

pair_freq = defaultdict(int)

for word in corpus:
    for i in range(len(word) - 1):
        pair = (word[i], word[i + 1])
        pair_freq[pair] += 1
print(pair_freq)
```
